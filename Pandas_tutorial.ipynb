```python
# Cleaning Data - Missing values

df = pd.read_csv("raw_data.csv")
df.isnull()
df.isnull().sum() # number of missing values
df.dropna() # drop rows
df.dropna(axis = 1) # drop columns
df.fillna(0) # fills missing values with another value

cleaned_data = df.copy()
cleaned_data
mean_age = cleaned_data["age"].mean()
cleaned_data["age"] = cleaned_data["age"].fillna(mean_age)
cleaned_data

df.ffill() # forward fill
df.bfill() # backward fill
```

```python
# Handling Strings

df["gender"].str.lower()
df["gender"].str.upper()
df["gender"].str.capitalize()
df["name"].str.split(" ")
df["country"].str.contains("india") # India -> False
df["country"].str.contains("india", case = False) # India -> True
```

```python
# Feature Engineering

df2 = df.copy()
# Apply
df2["tax"] = df2["income"].apply(lambda x : "20%" if x >= 60000 else "10%")

# Map
gender_map = {"Male" : "M", "Female" : "F", "Unknown": "U"}
df2["gender"] = df2["gender"].map(gender_map)

# Assign
df2 = df2.assign(new_income = df2["income"] * 1.1)

# Replace
df2["country"] = df2["country"].replace("USA","US")

#Rename
df2.columns = ["Id", "Name", "Age", "Country", "Gender", "Income", "Tax", "New_Income"]
df2.rename(columns ={"Income" : "Salary"})
df2.rename(index={1: "First"})

# Sort Values
# On the basis of columns
df2.sort_values("Income", ascending = False)
sorted_data = df2.sort_values(["Income", "Age"]) # Sorting is based on 1st parameter
sorted_data
# On the basis of index
sorted_data.sort_index()

# Reset index but sort the columns
sorted_data.reset_index()


# Ranking
# sorted_data["Ranking"] = sorted_data["Income"].rank()
# sorted_data["Ranking"] = sorted_data["Income"].rank(ascending = False)
sorted_data["Ranking"] = sorted_data["Income"].rank(ascending = False, method = "dense") # min and max -> other methods
sorted_data[["Id", "Name", "Age", "Country", "Gender", "Income", "New_Income", "Tax"]]
```
