```python
# Grouping and Aggregation of data

df2 = df.copy()
df2.groupby("country")["income"].mean()

df2.groupby("gender")["income"].agg(["mean","min","max"])

# Renaming columns
df2.groupby("gender")["income"].agg( avg_salary = "mean", min_salary= "min", max_salary = "max")

# Agg() on different columns

df2.groupby("country").aggregate({
    "income" : "min",
    "age" : "mean"
})

# Naming columns
df2.groupby("country").aggregate(
    min_salary = ("income", "min"),
    avg_age = ("age", "mean")
)
```

```python
# Melt 

df = pd.DataFrame({
    "country" : ["USA", "USA", "India", "India"],
    "year" : [2020, 2021, 2020, 2021],
    "sales" :[ 100, 120, 90, 150],
    "profit" : [20, 25, 21, 22],
    "tax" : [10, 20, 30, 40]
})

melted_df = df.melt(
    id_vars = ["country", "year"],
    value_vars = ["sales", "profit", "tax"],
    var_name = "metrics",
    value_name = "value"
)
melted_df

# Pivot
melted_df.pivot(
    index=["country","year"],
    columns="metrics",
    values="value"
)
```

```python
# Merging and Joining Data

df_customer = pd.DataFrame({
    "customer_id" : [2, 4, 5, 1],
    "name" : ["Bob", "Alice", "Rohan", "Joey"]
})

df_orders = pd.DataFrame({
    "order_id" : [312, 222, 456, 120],
    "customer_id" : [4, 1, 6, 3],
    "amount" : [250, 320, 180, 500]
})

# Inner Join
pd.merge(df_customer, df_orders, on="customer_id")

# Left Join
pd.merge(df_customer, df_orders, on="customer_id", how="left")

# Right Join
pd.merge(df_customer, df_orders, on="customer_id", how="right")

# Outer Join
pd.merge(df_customer, df_orders, on="customer_id", how="outer")
```

```python
# Data Concatenation

df1 = pd.DataFrame({
    "id" : [1, 2, 3],
    "name" : ["Haley", "Alex","Luke"]
})

df2 = pd.DataFrame({
    "id" : [4, 5, 6],
    "name" : ["Manny", "Lily", "Joe"]
})

pd.concat([df1, df2]) # Row wise
pd.concat([df1,df2], ignore_index=True) # Rewriting index values
pd.concat([df1, df2], axis = 1) # Column wise
```
