```python
import pandas as pd
```

```python
info = {
    "Name" : ["Anshika", "Kundan", "Shourya", "Shradha"],
    "CGPA" : [9.13, 9.15, 9.00, 9.3],
    "Gender" : ['F', 'M', 'M', 'F']
}

df = pd.DataFrame(info)

print(df)
```

```python
df
```

```python
# Series in Pandas

s = pd.Series([12, 3, 45, 18, 90])
print(s)
```

```python
# Defining the label of series

s = pd.Series([23, 34, 20], index = ["Ross", "Joey", "Chandler"])
print(s)
print(s["Ross"])
print(s["Chandler"])
print(s.index)
```

```python
# Vectorizede Operations

s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([10, 20, 30, 40, 50])
#print(s1+s2)
s1[0] = 100
print(s1)
change_s1 = s1.drop(0)
print(change_s1)
```

```python
# DataFrame creation

#Using Dictionary
info = {
    "Name" : ["Adam", "Eve", "Bob"],
    "Age" : [20, 23, 21],
    "Cgpa" : [8.35, 9.00, 7.8]
}

df = pd.DataFrame(info)

df
```

```python
# List of Lists

df = pd.DataFrame([["Adam", 20], ["Eve", 23], ["Bob", 21]], columns = ["Nmae", "Age"])
df
```

```python
# Using np Arrays

import numpy as np

np_arr = np.array([[1,2,3],[4,5,6]])
df = pd.DataFrame(np_arr, columns = ["A", "B", "C"])
print(df)
```
