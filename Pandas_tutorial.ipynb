```python
# Series in Pandas

s = pd.Series([12, 3, 45, 18, 90])
print(s)
```

```python
# Defining the label of series

s = pd.Series([23, 34, 20], index = ["Ross", "Joey", "Chandler"])
print(s)
print(s["Ross"])
print(s["Chandler"])
print(s.index)
```

```python
# Vectorizede Operations

s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([10, 20, 30, 40, 50])
#print(s1+s2)
s1[0] = 100
print(s1)
change_s1 = s1.drop(0)
print(change_s1)
```
